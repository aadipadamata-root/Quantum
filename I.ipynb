```python
from qiskit import QuantumCircuit, transpile

# 1. Build a simple 1-qubit circuit
qc = QuantumCircuit(1)
qc.h(0)
qc.measure_all()

# 2. Pick a healthy physical qubit (e.g., Qubit 1) to avoid the bad one (Qubit 0)
safe_qubit = 1

print(f"Mapping our virtual qubit 0 to a known safe physical qubit: {safe_qubit}")

# 3. Transpile using the standard Qiskit function by passing a simple list layout
compiled_circuit = transpile(
    qc, 

    initial_layout=[safe_qubit]  # 🎯 This forces Qiskit to use the safe physical qubit!
)

print("Circuit compiled successfully without complex custom passes!")
```
